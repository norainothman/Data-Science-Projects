```python
# We can also use dtype for Conversion
arr = np.array([10,20,30,40])
print(arr.dtype)
print(arr.nbytes) # total bytes used by an array

# int64 means 64 Bits
# means each and every element an array take upto 64 bits
# We have 4 elements
# 4 * 64 Bits --> Total size of arr : 256 Bits
# If we convert it into bytes 4 element * 8 bytes : 32 bytes

# behind the scene each and every individual element will gonna take 64 digits space
# 6 bits 6*4 --> 24 bits
# Total Size of arr : 24 bits
```

```python
arr = np.array([10,20,30,40],dtype='i1')
print(arr.dtype)
print(arr.nbytes)

# int8 means 8 Bits
# means each and every element an array take upto 8 bits
# We have 4 elements
# 4 * 8 Bits --> Total size of arr : 32 Bits
# If we convert it into bytes 4 element * 1 bytes : 4 bytes

# behind the scene each and every individual element will gonna take 8 digits space

# When in array you are storing the age or marks
```

```python
age = np.array([34,78,56,21,90,100,120,17])
print(age.dtype)
print(age.nbytes)
```

```python
# Memory efficient (Save Space -RAM)
age = np.array([34,78,56,21,90,100,120,17],dtype='i1')
print(age.dtype)
print(age.nbytes)

# i1 --> 8 digits space
# 21 --> 16 digits space
# i4 --> 32 digits space
# i8 --> 64 digits space
```

```python
# copy() VS view() methods of NumPy

# copy()
# Create new independent array in a different memory location

org_array = np.array([10,20,30,40,50])
copy_array = org_array.copy()
print(org_array)
print(copy_array)

# if we make changes in copy array, it won't effect the original array

copy_array[0] = 100
print(org_array)
print(copy_array)
```

```python
# view()
# Representation of original array
org_array = np.array([1,2,3,4,5,6])
print(org_array)

copied_arr = org_array.view()
print(copied_arr)

# if we make changes in copied array, it will effect original array
copied_arr[0] = 100
print(org_array)
print(copied_arr)

# if we make changes in original array, it will effect copied array
org_array[4] = 400
print(org_array)
print(copied_arr)
```

```python
org_array = np.array([1,2,3,4,5,6])
copied_arr = org_array.view()

# shares_memory() --> return True if arrays share same memory location
print(np.shares_memory(org_array,copied_arr))
```

```python
# Sorting
# np.sort() by default it will sort in ascending order

price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
print(np.sort(price))
```

```python
# Descending Order
price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
print(np.sort(price)[::-1])
```

```python
# Filter and Searching in Array

price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
# Boolean Masking
price >=50
```

```python
# Here we applied filter [ShowMeOnlyTrueValues]
new = price[price >=50]
print(new)
```

```python
price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
# Show even num price
price%2==0
```

```python
new = price[price%2==0]
print(new)
```

```python
# & combines both condition
price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
print(price[(price%2==0)&(price>=50)])
```

```python
# np.where(expression)
# Its return Index Number
price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
new = np.where(price%2==0)
print(price[new])
```

```python
# & combines both condition
price = np.array([78,34,56,90,102,34,12,67,172,120,95,45])
new = np.where((price%2==0)&(price>50))
print(price[new])
```

```python
# SALARY CLASSIFICATION
salary = np.array([25000,40000,55000,70000,30000,90000])
# Replace salary with High if salary >= 60000
# Replace salary with Medium if salary >= 40000 and <= 60000
# Otherwise Low

# np.where(condition,value_if_True,Value_if_False)
result = np.where(
    salary >= 60000, 'High',
    np.where(salary >= 40000, 'Medium', 'Low')
)
result
```
