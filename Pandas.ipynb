```python
# Data Science Library --> PANDAS
import numpy as np
import pandas as pd
```

```python
# numpy --> array()
arr = np.array([10,20,30,40])
print(type(arr))
print(arr)
```

```python
# PANDAS
# - Series (1D Array)
# - DataFrame (2D Array)
```

```python
# PANDAS (SERIES)
# Series is 1D Labelled Array
# Series is like Dictionary + Array combo

s = pd.Series([10,20,30,40])
print(type(s))
print(s)
print(s.shape)
```

```python
# In NumPy array we cannot store mixed data types
arr = np.array([123,'John',4500.32,True])
print(arr)
```

```python
s = pd.Series([123,'John',4500.32,True])
print(s)
print(s.values)
```

```python
s = pd.Series([123,'John',4500.32,True], index = ['id','name','sal','status'])
print(s)
print(s['sal'])

#   123   'John'   4500.32    True
#    0      1        2         3
#   id     name     sal       status

# If you want to get Data Value by using Index Number use iloc [Index Location]
print(s.iloc[3])

# get() --> build-in function

print(s.get('name'))
```

```python
# Indexing / Slicing / For Loop

s = pd.Series([123,'John',4500.32,True], index = ['id','name','sal','status'])
print(s)

#   123   'John'   4500.32    True
#    0      1        2         3
#   id     name     sal       status

# Indexing in Series can be done by using [], loc[] or iloc[]

print('\nIndexing In Series')

print(s['sal'])          # [] for Labels
print(s.loc['name'])     # loc[] for Labels
print(s.iloc[3])         # iloc[] for indexNumbers

print('\nSlicing In Series')
print(s[0:3])              # 3 will be excluded (stopIndex)
print(s['name':'status'])  # if using Labels, end label included

print('\nPrinting Specific Info In Series')
print(s[['id','name','status']])   # s[['LabelName','LabelName']]
print(s[['id','name','status']].values)
```

```python
# Using For Loop
for i in s:
  print(i)
```

```python
for i in s.items():
  print(i)
```

```python
for k,v in s.items():
  print(k,v)
```
